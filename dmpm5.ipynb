```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

# Define the dataset with correct comma separators between rows
dataset = [
    ['milk','onion','eggs','yogurt'],
    ['oil','onion','eggs','yogurt'],
    ['milk','apple','eggs'],
    ['milk','onion','corn','yogurt'],
    ['corn','onion','ice cream','eggs']
]

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)
print("Encoded DataFrame:")
display(df.head())

# Apply the apriori algorithm
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)
print("\nfrequent itemsets (Apriori):")
display(frequent_itemsets)

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
print("\nassociation rules:")
display(rules)

# Apply the FP-growth algorithm
frequent_itemsets_fp = fpgrowth(df, min_support=0.6, use_colnames=True)
print("\nfrequent itemsets (FP-growth):")
display(frequent_itemsets_fp)
```
