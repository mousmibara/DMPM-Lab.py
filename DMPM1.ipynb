```python
import numpy as np
import pandas as pd
from statistics import mean , median , mode , variance , stdev

data =[10,20,30,40,50]

print("mean:",mean (data))
print("median:",median (data))
print("mode:",mode (data))
print("variance:",variance (data))
print("standard deviation:",stdev (data))
```

```python
data=[20,40,60,80]
n=len(data)

#calculate meaan
mean=sum(data)/n
print("mean:",mean)

#calculate median
data.sort()
if n%2==0:
    median = (data[n//2-1]+data[n//2])/2
else:
    median=data[n//2]
print("median:",median)

#calculate mode
count={}
for i in data :
  count [i] = count.get (i,0)+1
mode = max(count,key=count.get)
print("mode:",mode)

#calculate deviation
std=variance**0.5
print("standard deviation:",std)

#calculate variance
s=0
for i in data:
    s+=(mean-i)**2
variance=s/n
print("variance:",variance)
```
