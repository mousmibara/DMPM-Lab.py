```python
import kagglehub
path = kagglehub.dataset_download("uciml/iris")
import pandas as pd
import os
csv_file = os.path.join(path, "Iris.csv")
iris_df = pd.read_csv(csv_file)
display(iris_df.head())
print(iris_df.info())
display(iris_df.describe())

# Note: check if column names are capitalized (usually 'Species' and 'Id' in this dataset)
# Let's clean and separate features (X) and target (y) properly
if 'Id' in iris_df.columns:
    iris_df = iris_df.drop('Id', axis=1)

target_col = 'Species' if 'Species' in iris_df.columns else 'species'
display(iris_df[target_col].value_counts())

X = iris_df.drop(target_col, axis=1)
y = iris_df[target_col]

print("shape of features (X):", X.shape)
print("shape of target (y):", y.shape)
print("\nfirst 5 rows of features (X):")
display(X.head())
print("\nfirst 5 rows of target (y):")
display(y.head())
```
